### Здесь проверяется работа простой функции математических операций.

```python
def test(number): # создаём функцию проверки
  return number * 2

print(test(5)) # выводим результат
```

### Пояснение
Функция `test()` принимает заданное число и умножает его на два.

### Ошибочный случай: передаём текст вместо числа

```python
try:
  print(test(abc)) # вводим числовое значение
except Exception as error:
  print("Ошибка:", error) # выводим ошибку
```

### Пояснение
Python передаётся текстовое значение вместо числового. Операция не может быть выполнена, и программа передаёт ошибку.
